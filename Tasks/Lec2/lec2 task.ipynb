3. Check if the season is Autumn, Winter, Spring or Summer.
    - If the user input is:
        - September, October or November, the season is Autumn.
        - December, January or February, the season is Winter.
        - March, April or May, the season is Spring
        - June, July or August, the season is Summer


````python
month = input("Enter the month: ").lower()
if month in ["september", "october", "november"]:
    print ("The season is Autumn.")
elif month in ["december", "january", "february"]:
    print ("The season is Winter.")
elif month in ["march", "april", "may"]:
    print ("The season is Spring.")
elif month in ["june", "july", "august"]:
    print ("The season is Summer.")
else:
    print ("Invalid month.")
````

```python

    'last_name': 'Parmar',
    'age': 96,
```

## Tasks

### 1. Check for `skills` key

### 2. Check for `'Python'` in skills
- Still inside the same condition:
  - Check if `'Python'` is in the person's skills list.
  - Print `True` if found, otherwise `False`.

### 3. Based on the combination of skills:
  - If the person has **only** `['Python', 'Matlab']` → print: `He knows machine learning`
  - If the person has both `'Python'` and `'R'` → print: `He knows statistics`
  - If the person has both `'C'` and `'C++'` → print: `He knows software development`

### 4. Print location and marital status
- If the person is married **and** lives in **Finland**, print:
```py

```



````python
#1
person = {
    'first_name': 'Milaan',
    'last_name': 'Parmar',
    'age': 96,
    'country': 'Finland',
    'is_marred': True,
    'skills': ['Python', 'Matlab', 'R', 'C', 'C++'],
    'address': {
        'street': 'Space street',
        'zipcode': '02210'
    }
}

if 'skills' in person:
    skills = person['skills']
    middle_skill = len(skills)//2
    print("Middle skill:", skills[middle_skill])
````

````python
#3
skills = person['skills']

if set(skills) == {'Python', 'Matlab'}:
    print("He knows machine learning")
elif 'Python' in skills and 'R' in skills:
    print("He knows statistics")
elif 'C' in skills and 'C++' in skills:
    print("He knows software development")
else:
    print("unknown title")
````

1. Go to the data folder and use the **[countries.py]()** file. Loop through the countries and extract all the countries containing the word **`land`**.


2. Go to the data folder and use the **[countries_details_data.py]()** file. 
   1. What are the total number of languages in the data ( output:  112 )
   2. Find the ten most spoken languages from the data ( output:  English , French , Arabic )
   3. Find the 10 most populated countries in the world ( output: China, India, US )

````python
from countries import countries
lands = []
for country in countries:
    if 'land' in country.lower():
        lands.append(country)
print(lands)
````

````python
from countries_details_dat import countries_data
#Q1
total_languages = set()
for country in countries_data:
    for language in country['languages']:
        total_languages.add(language)
print("Number of languages:", len(total_languages))
````

````python
#Q2
from collections import Counter

language_count = Counter()
for country in countries_data:
    for language in country['languages']:
        language_count[language] += 1
most_spoken = [lang for lang, count in language_count.most_common(10)]
print("10 Most spoken languages:", most_spoken)
````

````python
sorted_countries = sorted(countries_data, key=lambda c: c['population'], reverse=True)
most_populated = [country['name'] for country in sorted_countries[:10]]
print("10 most populated countries:", most_populated)
````
