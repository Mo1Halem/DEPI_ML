````python
#Q1
def area_of_circle(radius):
    pi = 3.14159
    return pi * radius ** 2

def perimeter_of_circle(radius):
    pi = 3.14159
    return 2 * pi * radius

print("Area of the circle =", area_of_circle(2))
print("Perimeter of the circle =", perimeter_of_circle(5))
````

````python
#Q2
def add_all_nums(*nums):
    total = 0
    for num in nums:
        total += num
    return total

print (add_all_nums(1, 2, 3, 4))
````

````python
#Q3
def convert_celsius_2_fahrenheit(celsius):
    return (celsius * 9/5) + 32

print("Temperature in Fahrenheit =", convert_celsius_2_fahrenheit(10))
````

````python
#Q4
def check_season(month):
    month = month.lower()
    if month in ["september", "october", "november"]:
        return "Autumn"
    elif month in ["december", "january", "february"]:
        return "Winter"
    elif month in ["march", "april", "may"]:
        return "Spring"
    elif month in ["june", "july", "august"]:
        return "Summer"
    else:
        return "Invalid month."

print(check_season("april"))
````

````python
#Q5
def calculate_slope(x1, y1, x2, y2):
    if x2 - x1 == 0:
        return "undefined."
    return (y2 - y1) / (x2 - x1)

print("Slope of the line =", calculate_slope(1, 2, 3, 4))
````

````python
#Q6
import math
import cmath    ## مستخدم الاتنين في حالة الreal او complex roots
def solve_quadratic_eqn(a, b, c):
    if a == 0:
        return "Not a quadratic equation."
    sqrt_eq = (b**2) - (4*a*c)
    if sqrt_eq >= 0:
        root1 = (-b + math.sqrt(sqrt_eq)) / (2*a)
        root2 = (-b - math.sqrt(sqrt_eq)) / (2*a)
    else:
        root1 = (-b + cmath.sqrt(sqrt_eq)) / (2*a)
        root2 = (-b - cmath.sqrt(sqrt_eq)) / (2*a)
    return root1, root2

print ("Roots of the quadratic equation (real roots)=", solve_quadratic_eqn(1, -4, 4))       #في حالة الreal roots
print ("Roots of the quadratic equation (complex roots)=", solve_quadratic_eqn(1, 2, 5))     #في حالة الcomplex roots   
````

````python
#Q7
def print_list(my_list):
    for item in my_list:
        print(item)

print_list([1, 3, "end"])
````

````python
#Q8
def capitalize_list(my_list):
    return [item.capitalize() for item in my_list]

print(capitalize_list(["capital", "test", "word"]))
````

````python
#Q9
def add_item_to_list(my_list, item):
    my_list.append(item)
    return my_list

food_staff = ['Potato', 'Tomato', 'Mango', 'Milk']
print(add_item_to_list(food_staff, 'Apple'))
````

````python
#Q10
def remove_item(my_list, item):
    if item in my_list:
        my_list.remove(item)
    return my_list

food_staff = ['Potato', 'Tomato', 'Mango', 'Milk']
print(remove_item(food_staff, 'Tomato'))
````

````python
#Q11
def sum_of_numbers(num):
    return sum(range(num + 1))

print(sum_of_numbers(5))
print(sum_of_numbers(10))
print(sum_of_numbers(100))
````

````python
#Q12
def sum_of_odds(num):
    return sum(number for number in range(num + 1) if number % 2 != 0)

print(sum_of_odds(5))
print(sum_of_numbers(10))
````

````python
#Q13
def sum_of_evens(num):
    return sum(number for number in range(num + 1) if number % 2 == 0)

print(sum_of_evens(5))
print(sum_of_evens(10))
````
